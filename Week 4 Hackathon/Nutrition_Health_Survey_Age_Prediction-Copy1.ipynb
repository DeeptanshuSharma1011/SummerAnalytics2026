```python
train_fe['age_group'] = train_fe['age_group'].map({
    'Adult':0,
    'Senior':1
})
```

```python
def bmi_category(bmi):
    if bmi < 18.5:
        return 0      
    elif bmi < 25:
        return 1      
    elif bmi < 30:
        return 2      
    else:
        return 3      

train_fe['BMI_Category'] = train_fe['BMXBMI'].apply(bmi_category)
test_fe['BMI_Category'] = test_fe['BMXBMI'].apply(bmi_category)
```

```python
train_fe['High_Glucose'] = (train_fe['LBXGLU'] >= 126).astype(int)

test_fe['High_Glucose'] = (test_fe['LBXGLU'] >= 126).astype(int)
```

```python
insulin_threshold = train_fe['LBXIN'].median()

train_fe['High_Insulin'] = (
    train_fe['LBXIN'] > insulin_threshold
).astype(int)

test_fe['High_Insulin'] = (
    test_fe['LBXIN'] > insulin_threshold
).astype(int)
```

```python
train_fe['Diabetes'] = (train_fe['DIQ010'] == 1).astype(int)

test_fe['Diabetes'] = (test_fe['DIQ010'] == 1).astype(int)
```

```python
train_fe['BMI_Activity'] = (
    train_fe['BMXBMI'] * train_fe['PAQ605']
)

test_fe['BMI_Activity'] = (
    test_fe['BMXBMI'] * test_fe['PAQ605']
)
```

```python
train_fe['Glucose_BMI'] = (
    train_fe['LBXGLU'] * train_fe['BMXBMI']
)

test_fe['Glucose_BMI'] = (
    test_fe['LBXGLU'] * test_fe['BMXBMI']
)
```

```python
train_fe['Glucose_Insulin_Ratio'] = (
    train_fe['LBXGLU'] /
    (train_fe['LBXIN'] + 1)
)

test_fe['Glucose_Insulin_Ratio'] = (
    test_fe['LBXGLU'] /
    (test_fe['LBXIN'] + 1)
)
```

```python
train_fe['Is_Active'] = (
    train_fe['PAQ605'] == 1
).astype(int)

test_fe['Is_Active'] = (
    test_fe['PAQ605'] == 1
).astype(int)
```

```python
train_fe['BMI_GLU_INS'] = (
    train_fe['BMXBMI'] *
    train_fe['LBXGLU'] *
    train_fe['LBXIN']
)

test_fe['BMI_GLU_INS'] = (
    test_fe['BMXBMI'] *
    test_fe['LBXGLU'] *
    test_fe['LBXIN']
)
```

```python
train_fe['BMI_Glucose_Ratio'] = (
    train_fe['BMXBMI'] /
    (train_fe['LBXGLU'] + 1)
)

test_fe['BMI_Glucose_Ratio'] = (
    test_fe['BMXBMI'] /
    (test_fe['LBXGLU'] + 1)
)
```

```python
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

from sklearn.model_selection import StratifiedKFold, cross_validate
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
scoring = {
    'accuracy': 'accuracy',
    'precision': 'precision',
    'recall': 'recall',
    'f1': 'f1',
    'roc_auc': 'roc_auc'
}
```

```python
models = {

    "Random Forest": RandomForestClassifier(
        n_estimators=500,
        max_depth=None,
        min_samples_split=5,
        min_samples_leaf=2,
        class_weight='balanced_subsample',
        random_state=42,
        n_jobs=-1
    ),

    "XGBoost": XGBClassifier(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=5,
        random_state=42,
        eval_metric='logloss'
    ),

    "LightGBM": LGBMClassifier(
        n_estimators=300,
        learning_rate=0.05,
        random_state=42
    ),

    "CatBoost": CatBoostClassifier(
        iterations=300,
        learning_rate=0.05,
        verbose=False,
        random_state=42
    ),

    "AdaBoost": AdaBoostClassifier(
        n_estimators=300,
        learning_rate=0.5,
        random_state=42
    )

}
```

```python
results = []

for name, model in models.items():

    scores = cross_validate(
        model,
        X,
        y,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

    results.append({

        "Model": name,

        "Accuracy Mean": scores['test_accuracy'].mean(),
        "Accuracy Std": scores['test_accuracy'].std(),

        "Precision Mean": scores['test_precision'].mean(),
        "Precision Std": scores['test_precision'].std(),

        "Recall Mean": scores['test_recall'].mean(),
        "Recall Std": scores['test_recall'].std(),

        "F1 Mean": scores['test_f1'].mean(),
        "F1 Std": scores['test_f1'].std(),

        "ROC-AUC Mean": scores['test_roc_auc'].mean(),
        "ROC-AUC Std": scores['test_roc_auc'].std()

    })
```

```python
comparison = pd.DataFrame(results)

comparison = comparison.sort_values(
    by="ROC-AUC Mean",
    ascending=False
)

comparison
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.barplot(
    data=comparison,
    x="ROC-AUC Mean",
    y="Model"
)

plt.title("Model Comparison (ROC-AUC)")
plt.xlabel("Mean ROC-AUC")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=comparison,
    x="F1 Mean",
    y="Model"
)

plt.title("Model Comparison (F1 Score)")
plt.xlabel("Mean F1 Score")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=comparison,
    x="Recall Mean",
    y="Model"
)

plt.title("Model Comparison (Recall)")
plt.xlabel("Mean Recall")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=comparison,
    x="Accuracy Mean",
    y="Model"
)

plt.title("Model Comparison (Accuracy)")
plt.xlabel("Mean Accuracy")

plt.show()
```
